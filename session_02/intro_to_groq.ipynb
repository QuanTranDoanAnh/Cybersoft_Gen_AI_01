```python
import os
from dotenv import load_dotenv

load_dotenv()  # Load environment variables from .env file
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(
    model="openai/gpt-oss-20b",  # Specify the model you want to use
    groq_api_key=os.getenv("GROQ_API_KEY")
)
```

```python
from langchain_core.messages import HumanMessage, AIMessage

messages = [
    AIMessage(content="Bạn là một trợ lý AI"),
    HumanMessage(content="Python là gì?")
]

result = model.invoke(messages)
```

```python
result.content
```

```python
# Message history / context
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
```

```python
store = {} # Đây là dict lưu trữ các cuộc hội thoại theo session_id

def get_session_history(session_id: str) -> BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]
```

```python
with_message_history = RunnableWithMessageHistory(model, get_session_history)
```

```python
chat_session_3 = {
    "configurable": {
        "session_id": "session_3"
    }
}
```

```python
messages = [
    HumanMessage(content="Tôi là ai, và tôi làm gì?"),
]

response = with_message_history.invoke(
    messages, 
    chat_session_3
)
```

```python
response.content
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "Bạn là một trợ lý AI, bạn sẽ trả lời bằng {language}"),
        MessagesPlaceholder(variable_name="input"),
    ]
)

chain = prompt | model
```

```python
chain_mesage_history = RunnableWithMessageHistory(
    chain, 
    get_session_history,
    input_messages_key="input"
)
```

```python
chat_session_new = {
    "configurable": {
        "session_id": "chat_session_new"
    }
}
```

```python
response = chain_mesage_history.invoke(
    {
        "input": [
            AIMessage(content="Bạn là giảng viên chuyên dạy về lập trình, hãy trả lời một cách chuyên nghiệp."),
            HumanMessage(content="Python là gì?"),
        ],
        "language": "Tiếng Việt"        
    },
    config=chat_session_new
)
```
